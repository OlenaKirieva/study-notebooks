```python
import pandas as pd
import numpy as np
```

```python
df_1 = pd.read_csv('/content/test1.csv', encoding='cp1251', sep=';', index_col = 'Номер')
```

```python
df_1
```

```python
result = {'Фактори': ['Вплив', 'Уміння надихати', 'Інтелектуальна стимуляція', 'Індивідуальний підхід', 'Мотивація', 'Управління', 'Надання самостійності'],
          'Оцінка': [df_1['бал'][[1, 8, 15]].sum(), df_1['бал'][[2, 9, 16]].sum(), df_1['бал'][[3, 10, 17]].sum(),
                     df_1['бал'][[4, 11, 18]].sum(), df_1['бал'][[5, 12, 19]].sum(),
                     df_1['бал'][[6, 13, 20]].sum(), df_1['бал'][[7, 14, 21]].sum()]}
```

```python
df_result = pd.DataFrame(result)
```

```python
df_result
```

```python
df_result['Інтерпретація'] = df_result['Оцінка'].apply(
    lambda x: 'високі здібності' if x >= 9
    else 'помірні здібності' if x >= 5
    else 'низькі здібності'
)
```

```python
df_result
```

```python
import matplotlib.pyplot as plt

# Визначаємо колір для кожного рівня здібностей
colors = {
    'високі здібності': '#81A684',
    'помірні здібності': '#D6B878',
    'низькі здібності': '#C98278'
}

bar_colors = df_result['Інтерпретація'].map(colors)

# Будуємо діаграму
plt.figure(figsize=(8, 5))

plt.bar(
    df_result['Фактори'],
    df_result['Оцінка'],
    color=bar_colors,
    width=0.6
)

plt.title('Оцінка здібностей за факторами')
plt.xlabel('Фактори')
plt.ylabel('Оцінка')

# Горизонтальні лінії на рівнях 4 і 8
plt.axhline(y=4.5, color='#999999', linestyle='--', linewidth=1)
plt.axhline(y=8.5, color='#999999', linestyle='--', linewidth=1)

# Повертаємо назви факторів під кутом, щоб вони не накладалися
plt.xticks(rotation=45, ha='right')

# Межі оцінки
plt.ylim(0, 12)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

counts = df_result['Інтерпретація'].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    counts,
    labels=counts.index,
    autopct='%1.0f%%',
    startangle=90
)

plt.title('Розподіл рівнів здібностей')
plt.show()
```
